```python
colunas_para_remover = ['Cabin', 'PassengerId', 'Name', 'Ticket']

df_limpo = df.drop(columns=colunas_para_remover)

df_limpo.head()
```

```python
mediana_idade = df_limpo['Age'].median()
df_limpo['Age'] = df_limpo['Age'].fillna(mediana_idade)

mediana_tarifa = df_limpo['Fare'].median()
df_limpo['Fare'] = df_limpo['Fare'].fillna(mediana_tarifa)

print(df_limpo.isnull().sum())
```

```python
df_limpo['Sex'] = df_limpo['Sex'].map({
    'male': 0,
    'female': 1
})

df_limpo.head()
```

```python
df_encoded = pd.get_dummies(
    df_limpo,
    columns=['Embarked'],
    dtype=int
)

df_encoded.head()
```

```python
df_encoded['FamilySize'] = (
    df_encoded['SibSp'] +
    df_encoded['Parch'] +
    1
)

df_encoded[['SibSp', 'Parch', 'FamilySize']].head()
```

```python
df_encoded['IsAlone'] = (
    df_encoded['FamilySize'] == 1
).astype(int)

df_encoded[['FamilySize', 'IsAlone']].head()
```

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

df_encoded[['Age', 'Fare']] = scaler.fit_transform(
    df_encoded[['Age', 'Fare']]
)

df_encoded.head()
```
